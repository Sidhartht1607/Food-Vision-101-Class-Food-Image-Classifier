## Starting with the CNN model building

```python
tf.random.set_seed(42)
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(128,128,3)),
    tf.keras.layers.Rescaling(1./255.),
    tf.keras.layers.Conv2D(10,3,activation='relu'),
    tf.keras.layers.Conv2D(10,3,activation='relu'),
    tf.keras.layers.MaxPool2D(pool_size=2, padding= "valid"),
    tf.keras.layers.Conv2D(10,3,activation='relu'),
    tf.keras.layers.Conv2D(10,3,activation='relu'),
    
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(10,activation='softmax')
    
                           
])

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
              loss=tf.keras.losses.SparseCategoricalCrossentropy(),metrics=['accuracy'])

```

```python
model.summary()
```

```python
history_1=model.fit(train_data,epochs=5, validation_data=validation_data)
```

```python
from tensorflow.keras.callbacks import EarlyStopping
```

```python
early_stopping = EarlyStopping(monitor='val_loss',
                         min_delta=0,
                         patience=5,
                         verbose=0,
                         start_from_epoch=0,
                         restore_best_weights=True)
tf.random.set_seed(42)
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(128,128,3)),
    tf.keras.layers.Rescaling(1./255.),
    tf.keras.layers.Conv2D(64,3,activation='relu'),
    tf.keras.layers.Conv2D(48,3,activation='relu'),
    tf.keras.layers.MaxPool2D(pool_size=2, padding= "valid"),
    tf.keras.layers.Conv2D(28,3,activation='relu'),
    tf.keras.layers.Conv2D(28,3,activation='relu'),
    
    tf.keras.layers.GlobalMaxPooling2D(),
    tf.keras.layers.Dense(10,activation='softmax')
    
                           
])

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
              loss=tf.keras.losses.SparseCategoricalCrossentropy(),metrics=['accuracy'])
```

```python
model.summary()
```

```python
history_2=model.fit(train_data,epochs=40, validation_data=validation_data,callbacks = early_stopping)
```

```python
model.evaluate(test_data)
```

## Write up
Try to build a model that would be able to correctly predict the food when given an image of the food item. The model is currently trained on 10 classes instead of 101. Base line numbers would give us 10% since it is 10 classes. Took accuracy as the metric since the classes are balanced. There were 6000 training samples  and 1500 validation samples and 2500 test samples.

### Results:
Models created and their test_data accuracy  

1. Baseline model: 10%  CNN with early stopping: 50.52% with validation accuracy of 48.27%
2. efficientb0(freeze)+GlobalAveragePooling :90.24%  with validation accuracy of 84.13%
3. efficientb0(freeze)+GlobalAveragePooling+early_stopping : 89.63%  with validation accuracy of 85.27%
4. Same model as above efficientb0(unfreeze)+GlobalAveragePooling +early_stopping: 90.31% with validation accuracy of 85.53 

The validation figures are from the best epoch by val_loss.. The other validation results are somewhere below these values were between 83%-85%. Training on unfrozen layer did not change the accuracy measurably.

One of the most notable differences was using the efficientb0 model after using the CNN model that was built from scratch. The efficientb0 model + globalaveragepooling only had 12810 trainable parameters compared to the 48,986 trainable paramters in the CNN model with earlystopping.




what went wrong: 
we had used a flatten layer on a huge filter map with many channels and it gave us a 336,410 paramters. This train/validation accuracy gap of nearly 50 points.This was a mistake which would make it memorize without making any progress in generalization. The convolution layer is the best to learn the image data as it also records the positions and uses far fewer paramters to learn.  

Using a pooling layer like globalmaxpooling with 10 channels underfit to 17% and with 28 channels it gave the best output compared to the other scratch CNN models.  The Global Max Pooling was not the problem but the channel count there might have been too much loss in information during training getting 10 numbers to describe an image is not going ideal. 

Fine-tuning after unfreezing the efficientnetb0 did not help with the model training  measurably. Best val_loss was epoch 1 . It overfit the model during epoch 3 with the model validation accuracy about 13 points less than the training accuracy.

hypothesis:
The prime_rib and pork_chop gave 33 errors plus 15 the other way with about 20% all showing one way
french_toast had the worst recall with 80.8% . 49 errors are about 20% of all 230 errors in the model  one class pair out of 45 possible pairs accounting for a fifth of everything wrong.
a hypothesis was that prime_rib with green vegetables on the plate was most likely reason for the problems in the predictions but they were  disregarded after viewing of the images of correct predictions of the prime_rib.
confidence when it was correct gave with 0.90 mean 0.75 when it was wrong. and there was 71% of correct predictions above 0.9 versus 33% of errors.

Confusion matrix makes sense as both the pork-chop and prime-rib look very similar in colour and how they are visible in a plate. The colour only matters when how cooked it is. So the classes are very close to same.

Assumptions:
The validation set is carved from Food-101's training split, which the dataset authors deliberately left uncleaned and which contains some wrong labels. The test split was manually reviewed. This likely explains why test accuracy (90.3%) exceeds validation accuracy (85.5%), and it means the training signal itself is partly noisy.
The validation_split is a smaller data that has been carved out of the training split. We could try to use 224*224*3 pixel training for future scope to improve the model further but using 128*128 *3helped to train faster as well as using a simpler model like the b0 instead of B2.
